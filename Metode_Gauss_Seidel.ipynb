```python
import numpy as np

def gauss_seidel(A, b, x0, tol=1e-5, max_iter=100):
    n = len(b)
    x = x0.copy()

    for k in range(max_iter):
        x_old = x.copy()

        for i in range(n):
            # Hitung jumlah dari elemen selain diagonal utama
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            # Perbarui nilai x[i] langsung menggunakan nilai terbaru (Gauss-Seidel)
            x[i] = (b[i] - s) / A[i][i]

        # Cek kriteria berhenti (toleransi error)
        error = np.max(np.abs(x - x_old))
        print(f"Iterasi ke-{k+1}: x = {x}")

        if error < tol:
            print(f"Konvergen pada iterasi ke-{k+1}")
            return x

    print("Mencapai batas iterasi maksimum.")
    return x

# Contoh Matriks A dan Vektor b
A = np.array([[4, 1, -1], [2, 7, 1], [1, -3, 12]], dtype=float)
b = np.array([3, 19, 31], dtype=float)
x0 = np.zeros(len(b))

# Jalankan fungsi
hasil = gauss_seidel(A, b, x0)
print("Hasil akhir:", hasil)
```
